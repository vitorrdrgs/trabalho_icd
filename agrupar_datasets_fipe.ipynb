```python
import pandas as pd
```

```python
lista_dados_integrar = [
    './data/raw/fipe_historico_carros_2026_05.csv',
    './data/raw/fipe_historico_carros_2026_06.csv'
]

df_total = pd.read_csv("data/dados_fipe_tratados.csv")
```

```python
df_final = []
for item in lista_dados_integrar:
    df_temp = pd.read_csv(item)
    df_final.append(df_temp)

df_final = pd.concat(df_final, ignore_index=True)
```

```python
df_total['reference_date'] = pd.to_datetime(df_total['reference_date'], format='ISO8601')
df_final['reference_date'] = pd.to_datetime(df_final['reference_date'], format='ISO8601')
```

```python
df_final['year_of_reference'] = df_final['reference_date'].dt.year
df_final['month_of_reference'] = df_final['reference_date'].dt.month_name(locale='us')
df_final = df_final.rename(columns={'preco_brl': 'brl_price', 'ano': 'year'})
df_final = df_final.drop(columns=['reference_code', 'reference_month', 'vehicle_type', 'brand_code', 'model_code', 'year_model_code', 'fuel_code', 'codigo_fipe', 'autenticacao', 'preco_raw', 'mes_referencia_api', 'tipo_veiculo', 'sigla_combustivel'])
```

```python
df_total = pd.concat([df_total, df_final], ignore_index=True).reset_index(drop=True)
```

```python
df_total.drop(columns='Unnamed: 0', inplace=True)
```

```python
df_total.to_csv('./data/dados_fipe_tratados_final.csv', index=False)
```
